```python
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "/content/drive/MyDrive/datasets/raw/"
```

```python
bureau = pd.read_csv(DATA_PATH + "bureau.csv")

print("Bureau shape:", bureau.shape)
display(bureau.head())
```

```python
bureau.info()
```

```python
print("Number of columns:", len(bureau.columns))
print("\nColumns:")
for col in bureau.columns:
    print("-", col)
```

```python
print("Data type distribution:")
display(bureau.dtypes.value_counts())
```

```python
print("Duplicate rows:", bureau.duplicated().sum())
```

```python
print("Total rows:", len(bureau))
print("Unique SK_ID_CURR:", bureau["SK_ID_CURR"].nunique())
print("Unique SK_ID_BUREAU:", bureau["SK_ID_BUREAU"].nunique())
```

```python
print("SK_ID_CURR represents the applicant.")
print("SK_ID_BUREAU represents an individual bureau credit record.")
print("Rows per applicant:")

credits_per_customer = bureau.groupby("SK_ID_CURR").size()
display(credits_per_customer.describe())
```

```python
print("Number of applicants with each bureau-history count:")
display(credits_per_customer.value_counts().sort_index().head(30))
```

```python
app_train_ids = pd.read_csv(DATA_PATH + "application_train.csv", usecols=["SK_ID_CURR"])

bureau_ids = set(bureau["SK_ID_CURR"].unique())
application_ids = set(app_train_ids["SK_ID_CURR"])
ids_without_bureau = application_ids - bureau_ids

print("Applicants with NO bureau history:", len(ids_without_bureau))
print("Percentage:", f"{len(ids_without_bureau) / len(app_train_ids) * 100:.2f}%")
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": bureau.isna().sum(),
    "Missing_Percentage": bureau.isna().mean() * 100
})

missing_summary = missing_summary.sort_values("Missing_Percentage", ascending=False)

print("Missing-value summary:")
display(missing_summary)
```

```python
numeric_cols = bureau.select_dtypes(include=np.number).columns.tolist()

print("Number of numeric columns:", len(numeric_cols))
print("\nNumeric columns:")
for col in numeric_cols:
    print("-", col)
```

```python
categorical_cols = bureau.select_dtypes(include=["object"]).columns.tolist()

print("Number of categorical columns:", len(categorical_cols))
print("\nCategorical columns:")
for col in categorical_cols:
    print("-", col)
```

```python
print("Numeric descriptive statistics:")
display(bureau[numeric_cols].describe().T)
```

```python
print("Categorical value distributions:")
for col in categorical_cols:
    print("\n" + "=" * 60)
    print(col)
    print("=" * 60)
    display(bureau[col].value_counts(dropna=False).head(20))
```

```python
if "CREDIT_ACTIVE" in bureau.columns:
    print("CREDIT_ACTIVE distribution:")
    display(bureau["CREDIT_ACTIVE"].value_counts(dropna=False))
```

```python
if "CREDIT_TYPE" in bureau.columns:
    print("CREDIT_TYPE distribution:")
    display(bureau["CREDIT_TYPE"].value_counts(dropna=False).head(30))
```

```python
credit_amount_cols = [col for col in [
    "AMT_CREDIT_SUM", "AMT_CREDIT_SUM_DEBT", "AMT_CREDIT_SUM_LIMIT",
    "AMT_CREDIT_SUM_OVERDUE", "AMT_ANNUITY"
] if col in bureau.columns]

print("Credit amount columns:", credit_amount_cols)
display(bureau[credit_amount_cols].describe().T)
```

```python
overdue_cols = [col for col in ["AMT_CREDIT_SUM_OVERDUE", "CREDIT_DAY_OVERDUE"] if col in bureau.columns]

print("Overdue columns:", overdue_cols)
if overdue_cols:
    display(bureau[overdue_cols].describe().T)
```

```python
#Overdue magnitude
if "CREDIT_DAY_OVERDUE" in bureau.columns:
    print("CREDIT_DAY_OVERDUE distribution (including zeros):")
    display(bureau["CREDIT_DAY_OVERDUE"].describe())

    print("\nCREDIT_DAY_OVERDUE distribution (only overdue records):")
    display(bureau[bureau["CREDIT_DAY_OVERDUE"] > 0]["CREDIT_DAY_OVERDUE"].describe())
```

```python
#Debt information
debt_cols = [col for col in ["AMT_CREDIT_SUM", "AMT_CREDIT_SUM_DEBT", "AMT_CREDIT_SUM_LIMIT"] if col in bureau.columns]

print("Debt-related columns:", debt_cols)
display(bureau[debt_cols].describe().T)
```

```python
#Debt vs credit sanity check
if "AMT_CREDIT_SUM_DEBT" in bureau.columns and "AMT_CREDIT_SUM" in bureau.columns:
    debt_exceeds_credit = (bureau["AMT_CREDIT_SUM_DEBT"] > bureau["AMT_CREDIT_SUM"]).sum()

    print(f"Records where debt exceeds total credit sum: {debt_exceeds_credit}")
    print(f"Percentage: {debt_exceeds_credit / len(bureau) * 100:.2f}%")
```

```python
#Credit status per applicant
if "CREDIT_ACTIVE" in bureau.columns:
    status_per_customer = bureau.groupby("SK_ID_CURR")["CREDIT_ACTIVE"].nunique()

    print("Number of different CREDIT_ACTIVE statuses per applicant:")
    display(status_per_customer.value_counts().sort_index())
```

```python
#Overdue records
if "CREDIT_DAY_OVERDUE" in bureau.columns:
    overdue_records = (bureau["CREDIT_DAY_OVERDUE"] > 0).sum()

    print("Records with CREDIT_DAY_OVERDUE > 0:", overdue_records)
    print("Percentage:", f"{overdue_records / len(bureau) * 100:.2f}%")
```

```python
#Active credit records
if "CREDIT_ACTIVE" in bureau.columns:
    active_records = (bureau["CREDIT_ACTIVE"] == "Active").sum()

    print("Active bureau credits:", active_records)
    print("Percentage:", f"{active_records / len(bureau) * 100:.2f}%")
```

```python
#Credit type per applicant
if "CREDIT_TYPE" in bureau.columns:
    credit_types_per_customer = bureau.groupby("SK_ID_CURR")["CREDIT_TYPE"].nunique()

    print("Distribution of number of different credit types per applicant:")
    display(credit_types_per_customer.value_counts().sort_index())
```

```python
#Days info
days_cols = [col for col in [
    "DAYS_CREDIT", "DAYS_CREDIT_ENDDATE", "DAYS_ENDDATE_FACT", "DAYS_CREDIT_UPDATE"
] if col in bureau.columns]

print("Date-related columns:", days_cols)
display(bureau[days_cols].describe().T)
```

```python
for col in days_cols:
    sentinel_count = (bureau[col] == 365243).sum()
    print(f"{col} - count of 365243: {sentinel_count}")
```

```python
if "CNT_CREDIT_PROLONG" in bureau.columns:
    print("Credit prolongation statistics:")
    display(bureau["CNT_CREDIT_PROLONG"].describe())
```

```python
example_customer = credits_per_customer.idxmax()

print("Applicant with the most bureau records:", example_customer)
display(bureau[bureau["SK_ID_CURR"] == example_customer])
```

```python
print("========== BUREAU SUMMARY ==========")
print("Total bureau rows:", len(bureau))
print("Unique applicants:", bureau["SK_ID_CURR"].nunique())
print("Unique bureau credits:", bureau["SK_ID_BUREAU"].nunique())
print("Applicants without bureau history:", len(ids_without_bureau))
print("Numeric columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))
```
